# TASK 1

```python
my_columns = ['Name', 'Author', 'Year', 'Price']
data = [
    ["Python Crash Course", "Eric Matthes", 2019, 45.50],
    ["Clean Code", "Robert C. Martin", 2008, 50.00],
    ["Fluent Python", "Luciano Ramalho", 2022, 60.20],
    ["Designing Data-Intensive Applications", "Martin Kleppmann", 2017, 55.00],
    ["The Pragmatic Programmer", "Andrew Hunt", 2019, 48.30],
    ["Python for Data Analysis", "Wes McKinney", 2022, 52.40],
    ["Introduction to Algorithms", "Thomas H. Cormen", 2009, 85.00],
]
df = pd.DataFrame(data, columns=my_columns)
df
```

```python
average_price = df["Price"].mean()
print(average_price)
```

```python
books_after_2015 = df[df["Year"] > 2015]
books_after_2015
```

```python
df_sorted = df.sort_values(by="Price", ascending=True)
df_sorted
```

# TASK 4

```python
df = pd.read_csv("employees_projects.csv")
df
```

```python
hours_per_employee = df.groupby("Имя")["Часы"].sum().reset_index(name="Всего часов")
hours_per_employee
```

```python
hours_per_project = df.groupby("Проект")["Часы"].sum().reset_index(name="Всего часов")
hours_per_project
```

```python
top_employee = hours_per_employee.loc[hours_per_employee["Всего часов"].idxmax()]
top_employee
```

# TASK 5

```python
df = pd.read_csv("ticket_sales.csv")
df
```

```python
tickets_per_movie = df.groupby("Фильм")["Проданные билеты"].sum().reset_index(name="Всего билетов")
tickets_per_movie
```

```python
tickets_per_city = df.groupby("Город")["Проданные билеты"].sum().reset_index(name="Всего билетов")
tickets_per_city
```

```python
top_movie = tickets_per_movie.loc[tickets_per_movie["Всего билетов"].idxmax()]
top_movie
```
