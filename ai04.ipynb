```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

# TASK 1

```python
days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
temperature = [18, 20, 22, 21, 24, 26, 23]
```

```python
plt.figure(figsize=(10, 5))
plt.plot(
    days,
    temperature,
    marker="o",
    linestyle="--",
    color="blue",
    linewidth=2,
    markersize=8,
)
plt.title("Weekly temperature", fontsize=14)
plt.xlabel("Days", fontsize=12)
plt.ylabel("Temperature (°C)", fontsize=12)

for i, txt in enumerate(temperature):
    plt.annotate(
        txt,
        (days[i], temperature[i]),
        textcoords="offset points",
        xytext=(0, 10),
        ha="center",
        fontsize=10,
    )

plt.grid(True, linestyle=":", alpha=0.7)
plt.show()
```

# TASK 2

```python
languages = ["Python", "JavaScript", "Java", "C#", "C++"]
students = [35, 28, 22, 18, 15]

max_index = np.argmax(students)
colors = ['#4e79a7', '#59a14f', '#edc948', '#b07aa1', '#ff9da7']

fig_h, ax_h = plt.subplots(figsize=(12, 7))
bars_h = ax_h.barh(languages, students, color=colors)

bars_h[max_index].set_edgecolor('black')
bars_h[max_index].set_linewidth(2)

ax_h.set_xlabel("Number of students", fontsize=12, fontweight='bold')
ax_h.set_ylabel("Languages", fontsize=12, fontweight='bold')
ax_h.set_title("Students by Programing Language", fontsize=14, fontweight='bold')

for bar in bars_h:
    width = bar.get_width()
    ax_h.text(width + 0.5,
              bar.get_y() + bar.get_height() / 2,
              f'{width}',
              ha='left',
              va='center',
              fontsize=11)

ax_h.set_xlim(0, max(students) + 5)
ax_h.invert_yaxis()

plt.tight_layout()
plt.show()
```

```python
fig_v, ax_v = plt.subplots(figsize=(12, 7))

bars_v = ax_v.bar(languages, students, color=colors)

bars_v[max_index].set_edgecolor('black')
bars_v[max_index].set_linewidth(2)

ax_v.set_xlabel("Языки программирования", fontsize=12, fontweight='bold')
ax_v.set_ylabel("Количество студентов", fontsize=12, fontweight='bold')
ax_v.set_title("Вертикальное представление распределения студентов", fontsize=14, fontweight='bold')

for bar in bars_v:
    height = bar.get_height()
    ax_v.text(bar.get_x() + bar.get_width() / 2,
              height + 0.5,
              f'{height}',
              ha='center',
              va='bottom',
              fontsize=11)

ax_v.set_ylim(0, max(students) + 5)

plt.tight_layout()
plt.show()
```

# TASK 3

```python
study_hours = [1, 2, 2, 3, 4, 4, 5, 6, 7, 8]
scores = [45, 50, 55, 60, 65, 70, 74, 80, 88, 95]
ages = [18, 19, 18, 20, 21, 19, 22, 20, 23, 21]

fig = plt.figure(figsize=(7, 4.5), dpi=150)

ax1 = fig.add_axes([0.1, 0.12, 0.7, 0.78])

sc = ax1.scatter(study_hours, scores, c=ages, cmap="viridis", s=80)

cax = fig.add_axes([0.83, 0.15, 0.03, 0.7])
cbar = fig.colorbar(sc, cax=cax)
cbar.set_label("Age", fontsize=10)

ax2 = fig.add_axes([0.15, 0.57, 0.25, 0.28])
ax2.bar(range(1, 11), scores, color="#1f77b4", width=0.7)
ax2.set_title("Scores", fontsize=8)
ax2.tick_params(labelsize=6)
ax2.set_xticks([2, 4, 6, 8, 10])

ax1.annotate(
    "Highest score",
    xy=(8, 95),
    xytext=(6.2, 91),
    arrowprops=dict(arrowstyle="->", color="black", lw=1),
    fontsize=10,
)

ax1.set_xlabel("Study Hours", fontsize=10)
ax1.set_ylabel("Exam Score", fontsize=10)
ax1.set_title("Study Hours vs Exam Score", fontsize=12)
ax1.grid(True, linestyle="-", alpha=0.2)

plt.show()
```
